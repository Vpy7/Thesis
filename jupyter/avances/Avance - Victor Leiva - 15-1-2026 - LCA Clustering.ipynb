```python
codes = {
    "pso": [(1, 5)],
    "pga": [(1, 5), (23, 29)],
    "psa": [(1, 12)],
    "pea": [(1, 4), (10, 12), (20, 20)],
    "pmd": [(1, 22), (35, 35)],
    "psz": [(1, 2), (9, 14)],
    "pad": [(1, 2), (4, 11), (22, 25), (27, 29), (31, 33), (44, 44)],
    "pcd": [(1, 4), (7, 9), (12, 15), (19, 23), (25, 25), (27, 29), (39, 39)],
    "pod": [(1, 12)],
    "pal": [(1, 1)],
    "pmj": [(1, 1)],
    "psu": [(1, 11)],
    "pni": [(1, 2)]
}

replacements = {
    'pea001': 'pea001z',
    'pea002': 'pea002k',
    'pea003': 'pea003k'
}

# Generar lista original
question_list = [
    f"{code}{num:03d}"
    for code, ranges in codes.items()
    for start, end in ranges
    for num in range(start, end + 1)
]


# **SOLUCIÓN CORRECTA:**
# Opción A: Usando función auxiliar
def apply_replacements(text, rep_dict):
    for old, new in rep_dict.items():
        text = text.replace(old, new)
    return text

question_list = [apply_replacements(item, replacements) for item in question_list]
df_sintomas_disc[question_list]
```

```python
# Definir el mapa de códigos global
CODE_MAP = {
    0: "NO", 2: "SI", 3: "AV", 1: "AV",
    7: "NR", 77: "NR",
    8: "NA", 88: "NA", 
    9: "NS", 99: "NS"
}

def calculate_imc_categories(df, weight_col='pea002k', min_weight_col='pea003k', height_col='pea001z'):
    """
    Calcula el IMC basado en peso y estatura, y categoriza en las categorías especificadas.
    
    Parameters:
    -----------
    df : DataFrame
        DataFrame con los datos
    weight_col : str
        Columna del peso actual en kilos (default: 'pea002k')
    min_weight_col : str
        Columna del menor peso en kilos (default: 'pea003k')
    height_col : str
        Columna de la estatura en centímetros (default: 'pea001z')
    
    Returns:
    --------
    DataFrame con dos nuevas columnas IMC_pea002 e IMC_pea003, y sin las columnas originales
    """
    
    df = df.copy()
    
    # Verificar que las columnas existan
    required_cols = [weight_col, min_weight_col, height_col]
    missing_cols = [col for col in required_cols if col not in df.columns]
    
    if missing_cols:
        print(f"Advertencia: Columnas no encontradas: {missing_cols}")
        return df
    
    # Convertir la estatura a metros (está en cm)
    df['height_m'] = df[height_col] / 100
    
    # Calcular IMC para peso actual
    df['IMC_pea002'] = df[weight_col] / (df['height_m'] ** 2)
    
    # Calcular IMC para menor peso
    df['IMC_pea003'] = df[min_weight_col] / (df['height_m'] ** 2)
    
    # Definir categorías del IMC
    imc_categories = ['underweight', 'normal weight', 'overweight', 'obese']
    imc_bins = [0, 18.5, 25, 30, np.inf]
    
    # Categorizar IMC para peso actual
    df['IMC_pea002'] = pd.cut(
        df['IMC_pea002'], 
        bins=imc_bins, 
        labels=imc_categories, 
        right=False
    )
    
    # Categorizar IMC para menor peso
    df['IMC_pea003'] = pd.cut(
        df['IMC_pea003'], 
        bins=imc_bins, 
        labels=imc_categories, 
        right=False
    )
    
    # Eliminar columnas temporales y originales
    df = df.drop(columns=['height_m', weight_col, min_weight_col, height_col])
    
    return df


def apply_code_map(df, columns, code_map=CODE_MAP):
    """
    Aplica el mapeo de códigos a las columnas especificadas.
    
    Parameters:
    -----------
    df : DataFrame
        DataFrame con los datos
    columns : list
        Lista de columnas a mapear
    code_map : dict
        Diccionario de mapeo de códigos
    
    Returns:
    --------
    DataFrame con códigos mapeados
    """
    df = df.copy()
    
    for col in columns:
        if col in df.columns:
            # Aplicar el mapeo, manteniendo los valores que no están en el mapa
            df[col] = df[col].apply(lambda x: code_map.get(x, x) if pd.notnull(x) else x)
    
    return df


def perform_lca_clustering(df, question_list, id_column='id', n_components_range=range(2, 7)):
    """
    Realiza LCA clustering con todas las variables como categóricas.
    """
    warnings.filterwarnings('ignore')
    
    print("="*70)
    print("ANÁLISIS DE CLASES LATENTES (LCA)")
    print("="*70)
    
    # PASO 1: Crear DataFrame de trabajo
    work_df = df.copy()
    print(f"\n[1] Preparando datos: {len(work_df)} observaciones")
    
    # PASO 2: Procesar columnas de IMC (si existen)
    imc_required_cols = ['pea001z', 'pea002k', 'pea003k']
    
    # Verificar si tenemos columnas de IMC en question_list
    imc_cols_in_question_list = [col for col in imc_required_cols if col in question_list]
    
    if imc_cols_in_question_list and all(col in work_df.columns for col in imc_required_cols):
        print(f"[2] Procesando columnas de IMC...")
        print(f"    Transformando peso y estatura a categorías de IMC...")
        
        # Calcular y categorizar IMC
        work_df = calculate_imc_categories(work_df)
        
        # Actualizar question_list: eliminar columnas originales, agregar categorías
        for col in imc_required_cols:
            if col in question_list:
                question_list.remove(col)
        
        # Agregar nuevas columnas de IMC categorizadas
        imc_new_cols = ['IMC_pea002', 'IMC_pea003']
        for new_col in imc_new_cols:
            if new_col in work_df.columns:
                question_list.append(new_col)
        
        print(f"    ✓ IMC categorizado: {imc_new_cols}")
    
    # PASO 3: Aplicar mapeo de códigos
    print(f"\n[3] Aplicando mapeo de códigos...")
    
    # Verificar que las columnas existan
    existing_cols = [col for col in question_list if col in work_df.columns]
    missing_cols = [col for col in question_list if col not in work_df.columns]
    
    if missing_cols:
        print(f"    ⚠ Columnas no encontradas: {missing_cols}")
        question_list = existing_cols
    
    work_df = apply_code_map_without_imputation(work_df, question_list)
    print(f"    ✓ Mapeo aplicado a {len(question_list)} columnas")
    
    # VERIFICACIÓN CRÍTICA: Todas deben ser categóricas
    print(f"\n[4] Verificando tipos de datos...")
    non_categorical = []
    for col in question_list:
        if col in work_df.columns:
            dtype = work_df[col].dtype
            if dtype not in ['object', 'category']:
                non_categorical.append((col, dtype))
    
    if non_categorical:
        print(f"    ⚠ ADVERTENCIA: Columnas no categóricas encontradas:")
        for col, dtype in non_categorical:
            print(f"       {col}: {dtype}")
        print(f"    Estas columnas se forzarán a categóricas para LCA")
    
    # PASO 5: Análisis de valores faltantes
    print(f"\n[5] Análisis de valores faltantes (NaNs):")
    nan_df = work_df[question_list].copy()
    
    for col in nan_df.columns:
        nan_count = nan_df[col].isna().sum()
        if nan_count > 0:
            nan_percent = (nan_count / len(nan_df)) * 100
            print(f"    {col:<15}: {nan_count:>4} NaNs ({nan_percent:>6.2f}%)")
    
    total_nans = nan_df.isna().sum().sum()
    total_cells = nan_df.shape[0] * nan_df.shape[1]
    total_nan_percent = (total_nans / total_cells) * 100
    print(f"\n    Total: {total_nans:,} NaNs en {total_cells:,} celdas ({total_nan_percent:.2f}%)")
    
    # PASO 6: Preparar datos para LCA (todas como categóricas)
    print(f"\n[6] Preparando datos para LCA (todas variables como categóricas)...")
    
    cluster_df = work_df[question_list].copy()
    
    # Forzar todas las columnas a categóricas
    for col in cluster_df.columns:
        # Convertir a string para manejo uniforme
        cluster_df[col] = cluster_df[col].astype(str)
    
    # Codificar variables categóricas para análisis
    le_dict = {}
    encoded_data = []
    feature_names = []
    
    for col in cluster_df.columns:
        le = LabelEncoder()
        
        # Manejar valores NaN: tratarlos como categoría especial
        col_data = cluster_df[col].copy()
        col_data = col_data.fillna('MISSING')  # Marcar NaN explícitamente
        
        # Codificar
        try:
            encoded = le.fit_transform(col_data)
            encoded_data.append(encoded)
            feature_names.append(col)
            le_dict[col] = le
        except Exception as e:
            print(f"    ✗ Error codificando {col}: {e}")
    
    if not encoded_data:
        print("    ✗ No se pudieron codificar datos para LCA")
        return work_df
    
    # Crear matriz de características
    X = np.column_stack(encoded_data)
    print(f"    ✓ Datos preparados: {X.shape[0]} × {X.shape[1]}")
    print(f"    ✓ Columnas codificadas: {len(feature_names)}")
    
    # PASO 7: Realizar LCA
    print(f"\n[7] Realizando Análisis de Clases Latentes...")
    
    results = {}
    metrics = []
    
    for n_components in n_components_range:
        print(f"\n    k = {n_components} clases latentes...")
        
        try:
            # Para variables categóricas, podemos usar LatentClassAnalysis
            # o seguir con GaussianMixture con covarianza diagonal
            gmm = GaussianMixture(
                n_components=n_components,
                covariance_type='diag',  # Mejor para variables categóricas
                random_state=42,
                n_init=10,
                max_iter=500,
                tol=1e-3
            )
            
            clusters = gmm.fit_predict(X)
            
            # Almacenar resultados
            results[n_components] = clusters
            
            # Calcular métricas
            bic = gmm.bic(X)
            aic = gmm.aic(X)
            
            metrics.append({
                'k': n_components,
                'BIC': bic,
                'AIC': aic,
                'Convergió': gmm.converged_,
                'Iteraciones': gmm.n_iter_
            })
            
            print(f"      ✓ BIC: {bic:.2f}, AIC: {aic:.2f}")
            
            # Distribución de clusters
            unique, counts = np.unique(clusters, return_counts=True)
            print(f"      ✓ Distribución:")
            for cluster, count in sorted(zip(unique, counts)):
                percentage = (count / len(clusters)) * 100
                print(f"          Clase {cluster}: {count} ({percentage:.1f}%)")
            
        except Exception as e:
            print(f"      ✗ Error: {str(e)}")
            results[n_components] = None
    
    # PASO 8: Crear DataFrame de salida
    print(f"\n[8] Creando DataFrame de salida...")
    
    # DataFrame base con id y columnas originales (mapeadas)
    output_cols = [id_column] + question_list
    available_cols = [col for col in output_cols if col in work_df.columns]
    output_df = work_df[available_cols].copy()
    
    # Agregar columnas de clusters
    for n_components, clusters in results.items():
        if clusters is not None:
            col_name = f'lca_k{n_components}'
            output_df[col_name] = clusters
            print(f"     Añadida columna: {col_name}")
    
    # PASO 9: Resumen final
    print(f"\n" + "="*70)
    print("RESUMEN FINAL - LCA CON VARIABLES CATEGÓRICAS")
    print("="*70)
    
    print(f"\nDataFrame de salida:")
    print(f"  • Dimensiones: {output_df.shape[0]} filas × {output_df.shape[1]} columnas")
    print(f"  • Variables categóricas: {len(question_list)}")
    
    if metrics:
        metrics_df = pd.DataFrame(metrics)
        print(f"\nMétricas de modelos:")
        print(metrics_df.to_string(index=False))
        
        # Recomendar mejor modelo (menor BIC)
        valid_metrics = metrics_df[metrics_df['BIC'].notnull()]
        if not valid_metrics.empty:
            best_model = valid_metrics.loc[valid_metrics['BIC'].idxmin()]
            print(f"\n  → Modelo recomendado: k = {best_model['k']} (menor BIC)")
    
    print(f"\n" + "="*70)
    print("PROCESO COMPLETADO")
    print("="*70)
    
    return output_df

def apply_code_map_without_imputation(df, columns, code_map=CODE_MAP):
    """
    Aplica el mapeo de códigos a las columnas especificadas sin imputar valores NaN.
    """
    df_result = df.copy()
    
    for col in columns:
        if col in df_result.columns:
            # Solo aplicar el mapeo a valores no nulos
            df_result[col] = df_result[col].apply(
                lambda x: code_map.get(x, x) if pd.notnull(x) else x
            )
    
    return df_result

df=df_sintomas_disc.copy()
final_df=perform_lca_clustering(df, question_list, id_column='id')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency
from itertools import combinations

print("=" * 100)
print("ANÁLISIS ESTADÍSTICO ")
print("=" * 100)

# Configuración para visualización completa
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
np.set_printoptions(precision=3, suppress=True)

# Identificar columnas de cluster
cluster_cols = sorted([col for col in final_df.columns if 'lca_k' in col])
print(f"\nCOLUMNAS DE CLUSTER ENCONTRADAS: {cluster_cols}")
print(f"Total de observaciones: {len(final_df)}")

# Identificar TODAS las variables categóricas (excluyendo columnas de cluster, ID y IMC)
cat_vars = [col for col in final_df.columns 
            if col not in cluster_cols 
            and col != 'id' 
            and not col.startswith('IMC_')
            and final_df[col].dtype in ['object', 'category', 'string']]

print(f"TOTAL variables categóricas identificadas: {len(cat_vars)}")

# ============================================================================
# ANÁLISIS 1: DISTRIBUCIÓN DE CLUSTERS Y ESTADÍSTICAS BÁSICAS
# ============================================================================
print("\n" + "=" * 80)
print("1. ESTADÍSTICAS DE DISTRIBUCIÓN DE CLUSTERS POR CLUSTERIZACIÓN")
print("=" * 80)

# Crear DataFrame de resumen
resumen_clusters = pd.DataFrame()

for cluster_col in cluster_cols:
    cluster_stats = final_df[cluster_col].value_counts().sort_index()
    n_clusters = len(cluster_stats)
    
    # Calcular estadísticas detalladas
    for cluster_id in cluster_stats.index:
        cluster_data = final_df[final_df[cluster_col] == cluster_id]
        cluster_size = len(cluster_data)
        
        # Calcular estadísticas de completitud para este cluster
        missing_stats = []
        for var in cat_vars:
            if var in cluster_data.columns:
                missing_pct = cluster_data[var].isna().mean() * 100
                if missing_pct < 50:  # Solo considerar variables con menos del 50% missing
                    missing_stats.append(missing_pct)
        
        avg_missing = np.mean(missing_stats) if missing_stats else 100
        
        resumen_clusters = pd.concat([
            resumen_clusters,
            pd.DataFrame([{
                'Clusterizacion': cluster_col,
                'Cluster': cluster_id,
                'N_Observaciones': cluster_size,
                'Porcentaje_Total': (cluster_size / len(final_df)) * 100,
                'Missing_Promedio': avg_missing,
                'Completitud_Promedio': 100 - avg_missing
            }])
        ])

# Mostrar resumen consolidado
print("\nRESUMEN DE TODOS LOS CLUSTERS:")
for cluster_col in cluster_cols:
    cluster_data = resumen_clusters[resumen_clusters['Clusterizacion'] == cluster_col]
    total_obs = cluster_data['N_Observaciones'].sum()
    
    print(f"\n{cluster_col}:")
    print(f"  Número de clusters: {len(cluster_data)}")
    print(f"  Total observaciones: {total_obs}")
    print(f"  Tamaño promedio de cluster: {cluster_data['N_Observaciones'].mean():.0f}")
    print(f"  Rango de tamaños: {cluster_data['N_Observaciones'].min():.0f} - {cluster_data['N_Observaciones'].max():.0f}")
    print(f"  Completitud promedio: {cluster_data['Completitud_Promedio'].mean():.1f}%")

# ============================================================================
# ANÁLISIS 2: IDENTIFICACIÓN DE VARIABLES CLAVE POR CLUSTER
# ============================================================================
print("\n" + "=" * 80)
print("2. IDENTIFICACIÓN SISTEMÁTICA DE VARIABLES CLAVE")
print("=" * 80)

# Para cada clusterización, identificar las variables más discriminantes
variables_discriminantes_global = {}

for cluster_col in cluster_cols:
    print(f"\nAnalizando variables para {cluster_col}...")
    
    discriminacion_scores = []
    
    # Analizar TODAS las variables categóricas
    for var in cat_vars:
        if var in final_df.columns:
            try:
                # Crear tabla de contingencia
                contingency_table = pd.crosstab(final_df[var], final_df[cluster_col])
                
                # Calcular chi-cuadrado y Cramér's V
                chi2, p, dof, expected = chi2_contingency(contingency_table)
                n = contingency_table.sum().sum()
                min_dim = min(contingency_table.shape) - 1
                cramers_v = np.sqrt(chi2 / (n * min_dim)) if min_dim > 0 else 0
                
                # Calcular poder discriminante adicional
                discriminacion_total = 0
                for cluster_id in contingency_table.columns:
                    cluster_vals = contingency_table[cluster_id]
                    total_cluster = cluster_vals.sum()
                    
                    if total_cluster > 0:
                        # Calcular concentración en categorías principales
                        top_2_cats = cluster_vals.nlargest(2)
                        concentracion = top_2_cats.sum() / total_cluster
                        discriminacion_total += concentracion
                
                discriminacion_promedio = discriminacion_total / len(contingency_table.columns)
                
                discriminacion_scores.append({
                    'Variable': var,
                    'Cramers_V': cramers_v,
                    'p_value': p,
                    'Chi2': chi2,
                    'Discriminacion_Promedio': discriminacion_promedio,
                    'Significativo': p < 0.001
                })
                
            except Exception as e:
                continue
    
    # Convertir a DataFrame y ordenar
    discriminacion_df = pd.DataFrame(discriminacion_scores)
    
    # Filtrar solo variables significativas
    variables_significativas = discriminacion_df[discriminacion_df['Significativo']].copy()
    
    # Ordenar por Cramér's V descendente
    variables_significativas = variables_significativas.sort_values('Cramers_V', ascending=False)
    
    print(f"  Variables significativas encontradas: {len(variables_significativas)} de {len(cat_vars)}")
    
    if len(variables_significativas) > 0:
        # Guardar para uso posterior
        variables_discriminantes_global[cluster_col] = variables_significativas
        
        # Mostrar top 15 variables más discriminantes
        print(f"\n  Top 25 variables más discriminantes para {cluster_col}:")
        print("  " + "-" * 90)
        print("  Variable                                Cramér's V   p-value    χ²       Discriminación")
        print("  " + "-" * 90)
        
        for idx, row in variables_significativas.head(25).iterrows():
            print(f"  {row['Variable'][:40]:40} {row['Cramers_V']:.4f}     {row['p_value']:.2e}  {row['Chi2']:.0f}       {row['Discriminacion_Promedio']:.3f}")

# ============================================================================
# ANÁLISIS 3: PERFIL COMPLETO POR CLUSTER - VERSIÓN CON FRECUENCIAS ABSOLUTAS
# ============================================================================
print("\n" + "=" * 80)
print("3. PERFIL COMPLETO POR CLUSTER (ANÁLISIS CON 25 VARIABLES MÁS DISCRIMINANTES)")
print("=" * 80)

for cluster_col in cluster_cols:
    print(f"\n{'#' * 70}")
    print(f"PERFILES DETALLADOS PARA: {cluster_col}")
    print(f"{'#' * 70}")
    
    if cluster_col in variables_discriminantes_global:
        top_variables = variables_discriminantes_global[cluster_col].head(25)['Variable'].tolist()
    else:
        top_variables = cat_vars
    
    print(f"Analizando {len(top_variables)} variables clave...")
    
    # Tamaño total del dataset para cálculos globales
    total_dataset = len(final_df)
    
    perfiles_detallados = []
    
    for cluster_id in sorted(final_df[cluster_col].unique()):
        cluster_data = final_df[final_df[cluster_col] == cluster_id]
        cluster_size = len(cluster_data)
        
        perfil_cluster = {
            'Cluster': cluster_id,
            'Tamaño': cluster_size,
            'Porcentaje': (cluster_size / total_dataset) * 100,
            'Variables_Destacadas': []
        }
        
        for var in top_variables:
            if var in cluster_data.columns:
                # FRECUENCIAS ABSOLUTAS en el cluster
                cluster_counts = cluster_data[var].value_counts()
                # FRECUENCIAS RELATIVAS en el cluster
                cluster_dist = cluster_counts / cluster_size
                
                # FRECUENCIAS ABSOLUTAS globales
                global_counts = final_df[var].value_counts()
                # FRECUENCIAS RELATIVAS globales
                global_dist = global_counts / total_dataset
                
                # Filtrar categorías con al menos 5% en el cluster
                cluster_dist_filtrado = cluster_dist[cluster_dist >= 0.05]
                
                if len(cluster_dist_filtrado) == 0:
                    cluster_dist_filtrado = cluster_dist.head(3)
                
                caracteristicas_sobre = []
                caracteristicas_sub = []
                
                for cat, freq_cluster_rel in cluster_dist_filtrado.items():
                    freq_global_rel = global_dist.get(cat, 0)
                    
                    # Obtener conteos absolutos
                    count_cluster = cluster_counts.get(cat, 0)
                    count_global = global_counts.get(cat, 0)
                    
                    # Calcular diferencia
                    diferencia = freq_cluster_rel - freq_global_rel
                    
                    # Si la diferencia es significativamente positiva (>10%)
                    if diferencia > 0.10:
                        caracteristicas_sobre.append({
                            'Categoria': cat,
                            'Freq_Cluster_Rel': freq_cluster_rel,
                            'Count_Cluster': count_cluster,
                            'Freq_Global_Rel': freq_global_rel,
                            'Count_Global': count_global,
                            'Diferencia': diferencia,
                            'Tipo': 'SOBRE'
                        })
                    # Si la diferencia es significativamente negativa (< -10%)
                    elif diferencia < -0.10:
                        caracteristicas_sub.append({
                            'Categoria': cat,
                            'Freq_Cluster_Rel': freq_cluster_rel,
                            'Count_Cluster': count_cluster,
                            'Freq_Global_Rel': freq_global_rel,
                            'Count_Global': count_global,
                            'Diferencia': diferencia,
                            'Tipo': 'SUB'
                        })
                
                # Ordenar por valor absoluto de diferencia
                caracteristicas_sobre.sort(key=lambda x: abs(x['Diferencia']), reverse=True)
                caracteristicas_sub.sort(key=lambda x: abs(x['Diferencia']), reverse=True)
                
                todas_caracteristicas = caracteristicas_sobre + caracteristicas_sub
                
                if todas_caracteristicas:
                    max_dif_abs = max([abs(c['Diferencia']) for c in todas_caracteristicas], default=0)
                    
                    perfil_cluster['Variables_Destacadas'].append({
                        'Variable': var,
                        'Caracteristicas_Sobre': caracteristicas_sobre,
                        'Caracteristicas_Sub': caracteristicas_sub,
                        'Todas_Caracteristicas': todas_caracteristicas,
                        'Num_Sobre': len(caracteristicas_sobre),
                        'Num_Sub': len(caracteristicas_sub),
                        'Max_Diferencia_Abs': max_dif_abs
                    })
        
        # Ordenar variables por máxima diferencia absoluta
        perfil_cluster['Variables_Destacadas'].sort(
            key=lambda x: x['Max_Diferencia_Abs'],
            reverse=True
        )
        
        perfiles_detallados.append(perfil_cluster)
    
    # Mostrar resultados con AMBAS frecuencias (relativas y absolutas)
    for perfil in perfiles_detallados:
        print(f"\n  CLUSTER {perfil['Cluster']}: {perfil['Tamaño']} observaciones ({perfil['Porcentaje']:.1f}%)")
        print(f"  {'─' * 80}")
        
        if perfil['Variables_Destacadas']:
            print(f"  Variables distintivas:")
            
            for idx, var_info in enumerate(perfil['Variables_Destacadas'], 1):
                var_name = var_info['Variable']
                caracteristicas_sobre = var_info['Caracteristicas_Sobre']
                caracteristicas_sub = var_info['Caracteristicas_Sub']
                total_caracteristicas = len(caracteristicas_sobre) + len(caracteristicas_sub)
                
                if total_caracteristicas > 0:
                    print(f"\n    {idx:2d}. {var_name}:")
                    
                    # Mostrar categorías sobrerrepresentadas
                    if caracteristicas_sobre:
                        for i, carac in enumerate(caracteristicas_sobre, 1):
                            print(f"        SOBRE {i}: '{carac['Categoria']}' - "
                                  f"{carac['Freq_Cluster_Rel']:.0%} ({carac['Count_Cluster']}/{perfil['Tamaño']}) "
                                  f"vs {carac['Freq_Global_Rel']:.0%} ({carac['Count_Global']}/{total_dataset}) global, "
                                  f"+{abs(carac['Diferencia'])*100:.0f}pp")
                    
                    # Mostrar categorías subrepresentadas
                    if caracteristicas_sub:
                        for i, carac in enumerate(caracteristicas_sub, 1):
                            print(f"        SUB {i}: '{carac['Categoria']}' - "
                                  f"{carac['Freq_Cluster_Rel']:.0%} ({carac['Count_Cluster']}/{perfil['Tamaño']}) "
                                  f"vs {carac['Freq_Global_Rel']:.0%} ({carac['Count_Global']}/{total_dataset}) global, "
                                  f"-{abs(carac['Diferencia'])*100:.0f}pp")
        else:
            print(f"  No se encontraron variables distintivas (diferencia >10% con global)")
        
        print(f"  {'─' * 80}")

# ============================================================================
# ANÁLISIS 5: PATRONES COMBINADOS E INTERACCIONES ENTRE 25 VARIABLES CLAVE
# ============================================================================
print("\n" + "=" * 80)
print("5. ANÁLISIS DE PATRONES COMBINADOS E INTERACCIONES ENTRE 25 VARIABLES CLAVE")
print("=" * 80)

# Para cada clusterización, analizar combinaciones de variables
for cluster_col in cluster_cols:
    print(f"\n{'#' * 70}")
    print(f"ANÁLISIS DE PATRONES COMBINADOS PARA: {cluster_col}")
    print(f"{'#' * 70}")
    
    # Obtener las 25 variables más discriminantes
    if cluster_col in variables_discriminantes_global:
        top_vars = variables_discriminantes_global[cluster_col].head(25)['Variable'].tolist()
        # Mostrar las variables que se van a analizar
        print(f"\n  Variables analizadas (25 más discriminantes):")
        for i in range(0, 25, 5):
            grupo = top_vars[i:min(i+5, 25)]
            print(f"    {i+1:2d}-{min(i+5, 25):2d}: {', '.join(grupo)}")
    else:
        top_vars = cat_vars[:25]
    
    # Analizar combinaciones de 2 variables (pares)
    patrones_combinados = []
    n_clusters = final_df[cluster_col].nunique()
    
    print(f"\n  Analizando combinaciones entre las 25 variables...")
    print(f"  Total de combinaciones posibles: {(25*24)//2} pares")
    print(f"  Se evaluarán {n_clusters} clusters")
    
    # Progreso
    total_combinaciones = 0
    combinaciones_analizadas = 0
    
    # Calcular Cramér's V global para cada par de variables primero (optimización)
    print(f"  Calculando asociaciones globales...")
    cramers_global = {}
    
    for i in range(len(top_vars)):
        for j in range(i+1, len(top_vars)):
            var1, var2 = top_vars[i], top_vars[j]
            if var1 in final_df.columns and var2 in final_df.columns:
                try:
                    cross_table = pd.crosstab(final_df[var1], final_df[var2])
                    chi2, p, dof, expected = chi2_contingency(cross_table)
                    n = cross_table.sum().sum()
                    min_dim = min(cross_table.shape) - 1
                    cramers_v = np.sqrt(chi2 / (n * min_dim)) if min_dim > 0 else 0
                    cramers_global[(var1, var2)] = cramers_v
                except:
                    cramers_global[(var1, var2)] = 0
    
    print(f"  Analizando patrones por cluster...")
    
    # Para cada cluster, analizar las asociaciones
    for cluster_id in sorted(final_df[cluster_col].unique()):
        cluster_data = final_df[final_df[cluster_col] == cluster_id]
        cluster_size = len(cluster_data)
        
        # Solo analizar clusters con tamaño suficiente
        if cluster_size < 20:
            continue
            
        print(f"    Cluster {cluster_id} ({cluster_size} observaciones)...")
        
        # Analizar cada par de variables
        for i in range(len(top_vars)):
            for j in range(i+1, len(top_vars)):
                var1, var2 = top_vars[i], top_vars[j]
                
                if var1 not in cluster_data.columns or var2 not in cluster_data.columns:
                    continue
                
                try:
                    # Calcular Cramér's V para este cluster
                    cluster_cross = pd.crosstab(cluster_data[var1], cluster_data[var2])
                    
                    if cluster_cross.size == 0:
                        continue
                    
                    chi2_cluster, p_cluster, dof_cluster, expected_cluster = chi2_contingency(cluster_cross)
                    n_cluster = cluster_cross.sum().sum()
                    min_dim_cluster = min(cluster_cross.shape) - 1
                    
                    if min_dim_cluster == 0:
                        continue
                        
                    cramers_v_cluster = np.sqrt(chi2_cluster / (n_cluster * min_dim_cluster))
                    
                    # Obtener Cramér's V global para este par
                    cramers_v_global = cramers_global.get((var1, var2), 0)
                    
                    # Calcular diferencia absoluta
                    diferencia_abs = abs(cramers_v_cluster - cramers_v_global)
                    
                    # Solo considerar diferencias significativas
                    if diferencia_abs > 0.15:  # Umbral reducido para capturar más patrones
                        # Calcular si es más fuerte o más débil
                        if cramers_v_cluster > cramers_v_global:
                            tipo = "MÁS FUERTE"
                        else:
                            tipo = "MÁS DÉBIL"
                        
                        # Calcular fuerza de la asociación en el cluster
                        if cramers_v_cluster >= 0.5:
                            fuerza_cluster = "FUERTE"
                        elif cramers_v_cluster >= 0.3:
                            fuerza_cluster = "MODERADA"
                        elif cramers_v_cluster >= 0.1:
                            fuerza_cluster = "DÉBIL"
                        else:
                            fuerza_cluster = "MUY DÉBIL"
                        
                        # Almacenar patrón
                        patrones_combinados.append({
                            'Variables': f"{var1} × {var2}",
                            'Cluster': cluster_id,
                            'Tamaño_Cluster': cluster_size,
                            'Asociacion_Global': cramers_v_global,
                            'Asociacion_Cluster': cramers_v_cluster,
                            'Diferencia': cramers_v_cluster - cramers_v_global,
                            'Diferencia_Abs': diferencia_abs,
                            'Tipo_Cambio': tipo,
                            'Fuerza_Cluster': fuerza_cluster,
                            'p_valor': p_cluster
                        })
                        
                except Exception as e:
                    # Silenciosamente continuar con el siguiente par
                    continue
    
    # Mostrar resultados
    if not patrones_combinados:
        print(f"\n  No se encontraron patrones de interacción significativos.")
        print(f"  (Ninguna combinación mostró diferencia > 0.15 en Cramér's V)")
        continue
    
    print(f"\n  RESUMEN: Se encontraron {len(patrones_combinados)} patrones de interacción significativos")
    print(f"  (Diferencia > 0.15 en coeficiente Cramér's V)")
    
    # Agrupar por cluster para mejor presentación
    clusters_unicos = sorted(set([p['Cluster'] for p in patrones_combinados]))
    
    for cluster_id in clusters_unicos:
        patrones_cluster = [p for p in patrones_combinados if p['Cluster'] == cluster_id]
        
        # Ordenar por diferencia absoluta descendente
        patrones_cluster.sort(key=lambda x: x['Diferencia_Abs'], reverse=True)
        
        # Mostrar top 10 para este cluster
        top_n = min(10, len(patrones_cluster))
        
        print(f"\n  --- CLUSTER {cluster_id} ---")
        print(f"  {len(patrones_cluster)} patrones de interacción significativos encontrados")
        print(f"  (Mostrando los {top_n} más fuertes)")
        print(f"  {'─' * 60}")
        
        for idx, patron in enumerate(patrones_cluster[:top_n], 1):
            # Formatear nombres de variables (cortar si son muy largos)
            var_names = patron['Variables']
            if len(var_names) > 50:
                var_names = var_names[:47] + "..."
            
            # Determinar símbolo según tipo de cambio
            if patron['Tipo_Cambio'] == "MÁS FUERTE":
                simbolo = "↑"
            else:
                simbolo = "↓"
            
            print(f"  {idx:2d}. {simbolo} {var_names}")
            print(f"       Asociación: Cluster={patron['Asociacion_Cluster']:.3f} ({patron['Fuerza_Cluster']}) "
                  f"vs Global={patron['Asociacion_Global']:.3f}")
            print(f"       Diferencia: {patron['Diferencia']:+.3f} ({patron['Tipo_Cambio']})")
            print(f"       p-valor: {patron['p_valor']:.4f}")
            print()
    
    # Análisis agregado por tipo de cambio
    print(f"\n  ANÁLISIS AGREGADO PARA {cluster_col}:")
    print(f"  {'─' * 60}")
    
    # Contar patrones por tipo de cambio
    patrones_fuertes = [p for p in patrones_combinados if p['Tipo_Cambio'] == "MÁS FUERTE"]
    patrones_debiles = [p for p in patrones_combinados if p['Tipo_Cambio'] == "MÁS DÉBIL"]
    
    print(f"  Patrones con asociación MÁS FUERTE en algún cluster: {len(patrones_fuertes)}")
    print(f"  Patrones con asociación MÁS DÉBIL en algún cluster: {len(patrones_debiles)}")
    
    # Top 5 patrones con mayor diferencia (positiva y negativa)
    if patrones_combinados:
        print(f"\n  TOP 5 PATRONES CON MAYOR CAMBIO POSITIVO (más fuertes en cluster):")
        patrones_pos = sorted([p for p in patrones_combinados if p['Diferencia'] > 0], 
                             key=lambda x: x['Diferencia'], reverse=True)[:5]
        for idx, patron in enumerate(patrones_pos, 1):
            print(f"    {idx}. Cluster {patron['Cluster']}: {patron['Variables']}")
            print(f"       De {patron['Asociacion_Global']:.3f} a {patron['Asociacion_Cluster']:.3f} "
                  f"(Δ={patron['Diferencia']:.3f})")
        
        print(f"\n  TOP 5 PATRONES CON MAYOR CAMBIO NEGATIVO (más débiles en cluster):")
        patrones_neg = sorted([p for p in patrones_combinados if p['Diferencia'] < 0], 
                             key=lambda x: x['Diferencia'])[:5]
        for idx, patron in enumerate(patrones_neg, 1):
            print(f"    {idx}. Cluster {patron['Cluster']}: {patron['Variables']}")
            print(f"       De {patron['Asociacion_Global']:.3f} a {patron['Asociacion_Cluster']:.3f} "
                  f"(Δ={patron['Diferencia']:.3f})")
    
    # Análisis por fuerza de asociación en el cluster
    print(f"\n  DISTRIBUCIÓN POR FUERZA DE ASOCIACIÓN EN LOS CLUSTERS:")
    fuerzas = {}
    for patron in patrones_combinados:
        fuerza = patron['Fuerza_Cluster']
        if fuerza not in fuerzas:
            fuerzas[fuerza] = 0
        fuerzas[fuerza] += 1
    
    for fuerza, count in sorted(fuerzas.items(), key=lambda x: x[1], reverse=True):
        porcentaje = (count / len(patrones_combinados)) * 100
        print(f"    {fuerza}: {count} patrones ({porcentaje:.1f}%)")

print("\n" + "=" * 80)
print("FIN DEL ANÁLISIS DE PATRONES COMBINADOS")
print("=" * 80)
# ============================================================================
# ANÁLISIS 6: JERARQUÍA Y ESTABILIDAD DE CLUSTERS
# ============================================================================
print("\n" + "=" * 80)
print("6. ANÁLISIS DE JERARQUÍA Y ESTABILIDAD ENTRE CLUSTERIZACIONES")
print("=" * 80)

# Analizar cómo se relacionan los clusters entre diferentes niveles
if len(cluster_cols) > 1:
    print("\nEvolución jerárquica de clusters:")
    
    # Comparar cada par de clusterizaciones consecutivas
    for i in range(len(cluster_cols) - 1):
        col_actual = cluster_cols[i]
        col_siguiente = cluster_cols[i + 1]
        
        print(f"\n  {col_actual} → {col_siguiente}:")
        print("  " + "-" * 60)
        
        # Calcular matriz de transición en frecuencias absolutas
        tabla_conteo = pd.crosstab(final_df[col_actual], final_df[col_siguiente])
        
        # Calcular matriz de transición en porcentajes
        transicion_porcentaje = pd.crosstab(final_df[col_actual], 
                                           final_df[col_siguiente], 
                                           normalize='index')
        
        # Para cada cluster actual
        for cluster_actual in sorted(final_df[col_actual].unique()):
            # Obtener total de observaciones en este cluster
            total_obs = tabla_conteo.loc[cluster_actual].sum()
            
            # Obtener distribución de porcentajes para este cluster
            distribucion = transicion_porcentaje.loc[cluster_actual]
            
            # Ordenar por porcentaje descendente
            distribucion_ordenada = distribucion.sort_values(ascending=False)
            
            # Filtrar solo destinos con porcentaje > 0 (evitar mostrar ceros)
            distribucion_filtrada = distribucion_ordenada[distribucion_ordenada > 0]
            
            # Construir lista de transiciones
            transiciones = []
            total_acumulado = 0
            
            for cluster_dest, porcentaje in distribucion_filtrada.items():
                # Calcular frecuencia absoluta
                frecuencia = tabla_conteo.loc[cluster_actual, cluster_dest]
                
                # Formatear la transición
                transicion_str = f"{porcentaje:.1%} → {cluster_dest} ({frecuencia} obs)"
                transiciones.append(transicion_str)
                
                # Acumular porcentaje
                total_acumulado += porcentaje
                
                # Mostrar hasta cubrir 99.5% o máximo 5 transiciones
                if total_acumulado >= 0.995 or len(transiciones) >= 5:
                    break
            
            # Verificar si falta alguna parte de la distribución
            if total_acumulado < 0.995 and len(distribucion_filtrada) > len(transiciones):
                # Calcular el porcentaje restante
                porcentaje_restante = 1 - total_acumulado
                
                # Sumar las frecuencias restantes
                clusters_restantes = distribucion_filtrada.index[len(transiciones):]
                frecuencia_restante = sum([tabla_conteo.loc[cluster_actual, dest] 
                                          for dest in clusters_restantes])
                
                if porcentaje_restante > 0.001:  # Solo mostrar si > 0.1%
                    transiciones.append(f"{porcentaje_restante:.1%} → otros ({frecuencia_restante} obs)")
            
            # Imprimir la línea para este cluster
            print(f"  Cluster {cluster_actual} (total={total_obs} obs): {', '.join(transiciones)}")
        
        print("  " + "-" * 60)
        
        # Mostrar también una tabla resumen compacta
        print("\n  Resumen tabular:")
        print("  Clúster | Total obs | Distribución principal")
        print("  " + "-" * 50)
        
        for cluster_actual in sorted(final_df[col_actual].unique()):
            total_obs = tabla_conteo.loc[cluster_actual].sum()
            distribucion = transicion_porcentaje.loc[cluster_actual]
            
            # Obtener las 2 principales transiciones
            principales = distribucion.nlargest(2)
            
            if len(principales) == 2:
                dest1, pct1 = principales.index[0], principales.iloc[0]
                dest2, pct2 = principales.index[1], principales.iloc[1]
                obs1 = tabla_conteo.loc[cluster_actual, dest1]
                obs2 = tabla_conteo.loc[cluster_actual, dest2]
                
                print(f"  {cluster_actual:^7} | {total_obs:^9} | {pct1:.1%}→{dest1}({obs1}), {pct2:.1%}→{dest2}({obs2})")
            elif len(principales) == 1:
                dest, pct = principales.index[0], principales.iloc[0]
                obs = tabla_conteo.loc[cluster_actual, dest]
                print(f"  {cluster_actual:^7} | {total_obs:^9} | {pct:.1%}→{dest}({obs})")
# ============================================================================
# ANÁLISIS 7: SÍNTESIS Y RECOMENDACIONES FINALES
# ============================================================================
print("\n" + "=" * 80)
print("7. SÍNTESIS INTEGRAL")
print("=" * 80)

# Crear resumen evaluativo para cada clusterización
evaluacion_final = []

for cluster_col in cluster_cols:
    n_clusters = final_df[cluster_col].nunique()
    cluster_sizes = final_df[cluster_col].value_counts().values
    
    # Calcular métricas de calidad
    # 1. Balance de clusters
    size_std = cluster_sizes.std()
    size_mean = cluster_sizes.mean()
    balance_score = 1 - (size_std / size_mean) if size_mean > 0 else 0
    
    # 2. Poder discriminante (usando Cramér's V promedio de variables top)
    if cluster_col in variables_discriminantes_global:
        discriminacion_scores = variables_discriminantes_global[cluster_col]['Cramers_V'].head(20)
        discriminacion_score = discriminacion_scores.mean() if len(discriminacion_scores) > 0 else 0
    else:
        discriminacion_score = 0
    
    # 3. Estabilidad (coherencia interna)
    coherencia_scores = []
    for cluster_id in final_df[cluster_col].unique():
        cluster_data = final_df[final_df[cluster_col] == cluster_id]
        
        # Calcular coherencia para variables clave
        for var in cat_vars[:10]:
            if var in cluster_data.columns:
                frecuencias = cluster_data[var].value_counts(normalize=True)
                if not frecuencias.empty:
                    max_freq = frecuencias.iloc[0]
                    coherencia_scores.append(max_freq)
    
    coherencia_score = np.mean(coherencia_scores) if coherencia_scores else 0
    
    # 4. Puntuación combinada
    puntuacion_combinada = (balance_score * 1/3 + 
                           discriminacion_score * 1/3 + 
                           coherencia_score * 1/3)
    
    # Determinar recomendación
    if puntuacion_combinada > 0.7:
        recomendacion = "EXCELENTE"
    elif puntuacion_combinada > 0.5:
        recomendacion = "BUENA"
    elif puntuacion_combinada > 0.3:
        recomendacion = "ACEPTABLE"
    else:
        recomendacion = "LIMITADA"
    
    evaluacion_final.append({
        'Clusterizacion': cluster_col,
        'N_Clusters': n_clusters,
        'Balance': f"{balance_score:.3f}",
        'Discriminacion': f"{discriminacion_score:.3f}",
        'Coherencia': f"{coherencia_score:.3f}",
        'Puntuacion_Total': f"{puntuacion_combinada:.3f}",
        'Recomendacion': recomendacion
    })

# Crear DataFrame de evaluación
eval_df = pd.DataFrame(evaluacion_final)
print("\nEVALUACIÓN COMPARATIVA DE TODAS LAS CLUSTERIZACIONES:")
print("-" * 90)
print(eval_df.sort_values('Puntuacion_Total', ascending=False).to_string(index=False))
print("-" * 90)
```

## Definiciones de Estadísticas Utilizadas**

###  Cramér's V**
**Fórmula**: 

$V = \sqrt{\frac{\chi^2}{n \times \min(r-1, c-1)}}$


**Interpretación**:
- **0.00-0.10**: Asociación nula/muy débil
- **0.10-0.30**: Asociación débil
- **0.30-0.50**: Asociación moderada
- **0.50-0.70**: Asociación fuerte
- **0.70-1.00**: Asociación muy fuerte/perfecta

**Uso**: Mide qué tan fuertemente cada variable discrimina entre clusters.

### Valor-p (p-value)**
**Interpretación**:
- **< 0.001**: Evidencia estadística muy fuerte
- **< 0.01**: Evidencia fuerte
- **< 0.05**: Evidencia moderada
- **≥ 0.05**: Evidencia insuficiente

**Uso**: Valida significancia estadística de asociaciones.

### Estadístico Chi-cuadrado (χ²)**
**Fórmula**:

$\chi^2 = \sum \frac{(O_{ij} - E_{ij})^2}{E_{ij}}$

Donde $O_{ij}$ = frecuencias observadas, $E_{ij}$ = frecuencias esperadas.

**Uso**: Base para calcular Cramér's V.


###  Métricas de Calidad de Clustering**

| Métrica | Fórmula | Rango | Interpretación |
|---------|---------|-------|----------------|
| **Balance** | $1 - \frac{\sigma_{\text{tamaños}}}{\mu_{\text{tamaños}}}$ | 0-1 | Uniformidad en distribución |
| **Discriminación** | Promedio(Cramér's V top 20) | 0-1 | Poder separador de variables |
| **Coherencia** | Promedio(frecuencia modal) | 0-1 | Homogeneidad interna |
| **Puntaje Total** | $0.33\times\text{Balance} + 0.33\times\text{Discriminación} + 0.33\times\text{Coherencia}$ | 0-1 | Evaluación combinada |
