```python
def hdbscan_analysis(df, cols_key=None, min_cluster_size=10):
    df_copy = df.copy()
    
    # One-hot encoding (NaN se codifica como all zeros)
    encoder = OneHotEncoder(sparse_output=False, dtype=int, handle_unknown='ignore')
    X_encoded = encoder.fit_transform(df_copy)
    
    # HDBSCAN clustering
    clusterer = hdbscan.HDBSCAN(
        metric='hamming',
        min_cluster_size=min_cluster_size
    )
    labels = clusterer.fit_predict(X_encoded)
    df_copy.loc[:, 'cluster'] = labels
    
    print("Clusters encontrados y su tamaño:")
    print(pd.Series(labels).value_counts())
    
    mask = labels != -1
    if mask.sum() > 1 and len(set(labels[mask])) > 1:
        sil_score = silhouette_score(X_encoded[mask], labels[mask], metric='hamming')
        ch_score = calinski_harabasz_score(X_encoded[mask], labels[mask])
        db_score = davies_bouldin_score(X_encoded[mask], labels[mask])
        print(f"\nMétricas de evaluación (excluye al grupo -1 de outliers):")
        print(f"Silhouette Score: {sil_score:.3f}")
        print(f"Calinski-Harabasz Score: {ch_score:.3f}")
        print(f"Davies-Bouldin Score: {db_score:.3f}")
    else:
        print("\nNo se pueden calcular métricas: muy pocos clusters o filas.")
    
    if cols_key is not None:
        total_general = 1558
        all_dfs = []
    
        for col in cols_key:
            if col in df_copy.columns:
                counts = df_copy.groupby('cluster')[col].value_counts().unstack(fill_value=0)
                percentages_total = counts / total_general * 100
                percentages_cluster = counts.div(counts.sum(axis=1), axis=0) * 100
    
                merged = pd.concat([
                    counts.add_suffix('_count'),
                    percentages_total.add_suffix('_%_total'),
                    percentages_cluster.add_suffix('_%_cluster')
                ], axis=1)
                merged['question'] = col
                all_dfs.append(merged)
    
        if all_dfs:  
            base_prop_df = pd.concat(all_dfs)
            cols = list(base_prop_df.columns)
            cols.insert(0, cols.pop(cols.index('question')))
            base_prop_df = base_prop_df[cols]
        else:
            base_prop_df = pd.DataFrame() 
    
        return df_copy[['cluster']], base_prop_df

    else:
        return df_copy[['cluster']]


def min_cluster_search(full_dict, code, min_cluster_range):
    results = []

    for min_size in min_cluster_range:
        df_clusters, X_encoded = lca_mixed_analysis(
            full_dict=full_dict,
            code=code,
            min_cluster_size=min_size
        )

        labels = df_clusters['cluster'].values

        n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
        n_outliers = (labels == -1).sum()

        mask = labels != -1
        if mask.sum() > 1 and n_clusters > 0:
            sil_score = silhouette_score(X_encoded[mask], labels[mask], metric='hamming')
            ch_score = calinski_harabasz_score(X_encoded[mask], labels[mask])
            db_score = davies_bouldin_score(X_encoded[mask], labels[mask])
        else:
            sil_score = ch_score = db_score = None

        results.append({
            'min_cluster_size': min_size,
            'num_clusters': n_clusters,
            'num_outliers': n_outliers,
            'silhouette_score': sil_score,
            'calinski_harabasz_score': ch_score,
            'davies_bouldin_score': db_score
        })

    return pd.DataFrame(results)


def plot_hdbscan_search(search_results,code):

    fig = make_subplots(
        rows=3, cols=2,
        subplot_titles=(
            "Número de clusters",
            "Número de outliers",
            "Silhouette Score",
            "Calinski-Harabasz Score",
            "Davies-Bouldin Score",
            ""
        ),
        vertical_spacing=0.15,
        horizontal_spacing=0.1
    )

    # Número de clusters
    fig.add_trace(go.Scatter(
        x=search_results['min_cluster_size'],
        y=search_results['num_clusters'],
        mode='lines+markers',
        name='Número de clusters'
    ), row=1, col=1)

    # Número de outliers
    fig.add_trace(go.Scatter(
        x=search_results['min_cluster_size'],
        y=search_results['num_outliers'],
        mode='lines+markers',
        name='Número de outliers'
    ), row=1, col=2)

    # Silhouette Score
    fig.add_trace(go.Scatter(
        x=search_results['min_cluster_size'],
        y=search_results['silhouette_score'],
        mode='lines+markers',
        name='Silhouette Score'
    ), row=2, col=1)

    # Calinski-Harabasz Score
    fig.add_trace(go.Scatter(
        x=search_results['min_cluster_size'],
        y=search_results['calinski_harabasz_score'],
        mode='lines+markers',
        name='Calinski-Harabasz Score'
    ), row=2, col=2)

    # Davies-Bouldin Score
    fig.add_trace(go.Scatter(
        x=search_results['min_cluster_size'],
        y=search_results['davies_bouldin_score'],
        mode='lines+markers',
        name='Davies-Bouldin Score'
    ), row=3, col=1)

    fig.update_layout(
        title='Evaluación de min_cluster_size para HDBSCAN - '+full_dict[code]["nombre"] + " - " + code,
        template='plotly_white',
        height=900
    )

    fig.show()

def hdbscan_normalized_scores(df):
    ranked_df = df.copy()
    
    # Métricas a minimizar → la mejor = 1, la peor < 1
    ranked_df['norm_score_outliers'] = ranked_df['num_outliers'].min() / ranked_df['num_outliers']
    ranked_df['norm_score_db'] = ranked_df['davies_bouldin_score'].min() / ranked_df['davies_bouldin_score']
    
    # Métricas a maximizar → la mejor = 1, la peor < 1
    ranked_df['norm_score_silhouette'] = ranked_df['silhouette_score'] / ranked_df['silhouette_score'].max()
    ranked_df['norm_score_ch'] = ranked_df['calinski_harabasz_score'] / ranked_df['calinski_harabasz_score'].max()
    
    # Suma de puntajes normalizados
    ranked_df['norm_score_sum'] = (
        ranked_df['norm_score_outliers'] +
        ranked_df['norm_score_db'] +
        ranked_df['norm_score_silhouette'] +
        ranked_df['norm_score_ch']
    )
    
    return ranked_df.sort_values('norm_score_sum', ascending=False).head()

def plot_2d_pca(full_dict, code):
    df = pd.concat([full_dict[code]['df'], full_dict[code]['cluster_df']], axis=1)
    df = df[df['cluster'] != -1]

    X = df.filter(regex=f"^{code}").to_numpy()
    y = df["cluster"].to_numpy()

    mask = ~np.isnan(X)
    X_filled = np.where(mask, X, 0)

    n_components = 2
    max_iter = 100
    tol = 1e-6

    for i in range(max_iter):
        col_mean = np.nanmean(X_filled, axis=0)
        X_centered = X_filled - col_mean

        U, s, VT = np.linalg.svd(X_centered, full_matrices=False)
        X_recon = (U[:, :n_components] @ np.diag(s[:n_components]) @ VT[:n_components, :]) + col_mean

        diff = np.nan_to_num(X_filled - X_recon)
        X_filled[~mask] = X_recon[~mask]

        if np.linalg.norm(diff) < tol:
            break

    X_pca = U[:, :n_components] @ np.diag(s[:n_components])

    fig = px.scatter(
        x=X_pca[:,0],
        y=X_pca[:,1],
        color=y.astype(str),
        title=f'Clusters in 2D PCA Space - {full_dict[code]["nombre"]} - {code}',
        labels={'x':'PC1','y':'PC2','color':'cluster'}
    )
    fig.show()

def plot_3d_pca_subplots(full_dict, code):
    df = pd.concat([full_dict[code]['df'], full_dict[code]['cluster_df']], axis=1)
    df = df[df['cluster'] != -1]

    X = df.filter(regex=f"^{code}").to_numpy()
    y = df["cluster"].to_numpy()

    mask = ~np.isnan(X)
    X_filled = np.where(mask, X, 0)

    n_components = 3
    max_iter = 100
    tol = 1e-6

    for _ in range(max_iter):
        col_mean = np.nanmean(X_filled, axis=0)
        X_centered = X_filled - col_mean

        U, s, VT = np.linalg.svd(X_centered, full_matrices=False)
        X_recon = (U[:, :n_components] @ np.diag(s[:n_components]) @ VT[:n_components, :]) + col_mean

        diff = np.nan_to_num(X_filled - X_recon)
        X_filled[~mask] = X_recon[~mask]

        if np.linalg.norm(diff) < tol:
            break

    X_pca = U[:, :n_components] @ np.diag(s[:n_components])

    combos = [(0,1), (0,2), (1,2)]
    subplot_titles = [f'PC{i+1} vs PC{j+1}' for i,j in combos]

    fig = make_subplots(rows=1, cols=3, subplot_titles=subplot_titles)

    clusters_unique = np.unique(y)
    colors = px.colors.qualitative.Set2
    cluster_colors = {cluster: colors[i % len(colors)] for i, cluster in enumerate(clusters_unique)}

    for idx, (i,j) in enumerate(combos):
        for cluster in clusters_unique:
            mask_cluster = y == cluster
            scatter = go.Scatter(
                x=X_pca[mask_cluster,i],
                y=X_pca[mask_cluster,j],
                mode='markers',
                marker=dict(color=cluster_colors[cluster], opacity=0.5),
                name=f'Cluster {cluster}',
                showlegend=True if idx == 0 else False
            )
            fig.add_trace(scatter, row=1, col=idx+1)

        fig.update_xaxes(title_text=f"PC{i+1}", row=1, col=idx+1)
        fig.update_yaxes(title_text=f"PC{j+1}", row=1, col=idx+1)

    fig.update_layout(
        title_text=f'Clusters in 3D PCA Space - {full_dict[code]["nombre"]} - {code}',
        height=300,
        width=1000
    )
    fig.show()

def hdbscan_analysis_v2(full_dict, code, min_cluster_size=10):
    df = full_dict[code]['df']
    df_copy = df.copy()

    limited_cols = full_dict[code]['df_comp_pct_limited'].loc[
        full_dict[code]['df_comp_pct_limited']['limited'] == True, 'question'
    ].tolist()

    non_limited_cols = [c for c in df_copy.columns if c not in limited_cols]

    df_copy = df_copy.dropna(axis=1, how='all')

    cat_cols = [c for c in limited_cols if c in df_copy.columns]
    num_cols = [c for c in non_limited_cols if c in df_copy.columns]

    for c in cat_cols:
        df_copy[c] = df_copy[c].astype(str)
        df_copy[c] = df_copy[c].replace("nan", "__MISSING__")

    X_cat = pd.DataFrame(index=df_copy.index)
    if cat_cols:
        encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
        X_cat_arr = encoder.fit_transform(df_copy[cat_cols])
        X_cat = pd.DataFrame(
            X_cat_arr,
            columns=encoder.get_feature_names_out(cat_cols),
            index=df_copy.index
        )

    X_num = pd.DataFrame(index=df_copy.index)
    if num_cols:
        nan_flags = df_copy[num_cols].isna().astype(int)
        nan_flags.columns = [f"{c}_is_nan" for c in nan_flags.columns]

        X_num_filled = df_copy[num_cols].fillna(0)

        scaler = MinMaxScaler()
        X_num_scaled = pd.DataFrame(
            scaler.fit_transform(X_num_filled),
            columns=num_cols,
            index=df_copy.index
        )

        X_num = pd.concat([X_num_scaled, nan_flags], axis=1)

    X_encoded = pd.concat([X_num, X_cat], axis=1)

    if X_encoded.shape[0] < min_cluster_size:
        df_copy['cluster'] = -1
        return df_copy[['cluster']], X_encoded

    clusterer = hdbscan.HDBSCAN(
        metric='euclidean',
        min_cluster_size=min_cluster_size
    )

    labels = clusterer.fit_predict(X_encoded)
    df_copy['cluster'] = labels

    return df_copy[['cluster']], X_encoded

CODE_MAP = {
    0: "NO", 2: "SI", 3: "AV",
    7: "NR", 77: "NR",
    8: "NA", 88: "NA",
    9: "NS", 99: "NS"
}

def lca_mixed_analysis(full_dict, code, n_classes=6, max_iter=100):
    df = full_dict[code]["df"].copy()

    limited_cols = full_dict[code]["df_comp_pct_limited"].loc[
        full_dict[code]["df_comp_pct_limited"]["limited"] == True, "question"
    ].tolist()

    df = df.dropna(axis=1, how="all")

    cat_cols = [c for c in limited_cols if c in df.columns]
    num_cols = [c for c in df.columns if c not in cat_cols]

    for c in cat_cols:
        df[c] = df[c].map(CODE_MAP).fillna("MISSING")

    for c in num_cols:
        df[c] = pd.to_numeric(df[c], errors="coerce")
        df[c] = df[c].fillna(df[c].mean())

    scaler = StandardScaler()
    X_num = scaler.fit_transform(df[num_cols]) if num_cols else np.zeros((len(df), 0))

    cat_maps = {}
    X_cat = []
    for c in cat_cols:
        cats = df[c].astype("category").cat.categories
        cat_maps[c] = {k: i for i, k in enumerate(cats)}
        X_cat.append(df[c].map(cat_maps[c]).values)

    X_cat = np.vstack(X_cat).T if cat_cols else np.zeros((len(df), 0))

    n = len(df)
    K = n_classes
    Dn = X_num.shape[1]
    Dc = X_cat.shape[1]

    pi = np.ones(K) / K
    mu = np.random.randn(K, Dn)
    sigma = np.array([np.eye(Dn) for _ in range(K)])

    cat_probs = []
    for j in range(Dc):
        m = int(X_cat[:, j].max() + 1)
        cat_probs.append(np.ones((K, m)) / m)

    for _ in range(max_iter):
        log_resp = np.zeros((n, K))

        for k in range(K):
            if Dn > 0:
                diff = X_num - mu[k]
                inv = np.linalg.inv(sigma[k])
                log_resp[:, k] -= 0.5 * np.sum(diff @ inv * diff, axis=1)

            for j in range(Dc):
                log_resp[:, k] += np.log(cat_probs[j][k][X_cat[:, j]] + 1e-12)

            log_resp[:, k] += np.log(pi[k])

        log_resp -= log_resp.max(axis=1, keepdims=True)
        resp = np.exp(log_resp)
        resp /= resp.sum(axis=1, keepdims=True)

        Nk = resp.sum(axis=0)
        pi = Nk / n

        if Dn > 0:
            for k in range(K):
                mu[k] = (resp[:, k][:, None] * X_num).sum(axis=0) / Nk[k]
                diff = X_num - mu[k]
                sigma[k] = (resp[:, k][:, None, None] * np.einsum("ni,nj->nij", diff, diff)).sum(axis=0) / Nk[k]
                sigma[k] += np.eye(Dn) * 1e-6

        for j in range(Dc):
            for k in range(K):
                for c in range(cat_probs[j].shape[1]):
                    cat_probs[j][k, c] = resp[X_cat[:, j] == c, k].sum()
                cat_probs[j][k] /= Nk[k]

    labels = resp.argmax(axis=1)

    df = df.copy()
    df["cluster"] = labels

    lca_stats = pd.DataFrame({
        "cluster": range(K),
        "size": Nk.astype(int),
        "pi": pi
    })

    print("Clusterización final:")
    print(lca_stats)

    df_full = df.copy()
    df_full["pi_assigned"] = pi[labels]

    return df_full, {
        "pi": pi,
        "lca_stats": lca_stats,
        "mu": mu,
        "sigma": sigma,
        "cat_probs": cat_probs
    }


def lca_evaluate_clusters(full_dict, code, max_k=6, max_iter=100):
    df = full_dict[code]["df"].copy()

    # Columnas limited / non-limited
    limited_cols = full_dict[code]["df_comp_pct_limited"].loc[
        full_dict[code]["df_comp_pct_limited"]["limited"] == True, "question"
    ].tolist()
    all_cols = df.columns.tolist()
    count_limited = len([c for c in all_cols if c in limited_cols])
    count_non_limited = len(all_cols) - count_limited

    cluster_df = pd.DataFrame(index=df.index)
    rows = []

    for k in range(2, max_k + 1):
        print(f"{full_dict[code]['nombre']} - {code} - iter={k}")
        labels, model_params = lca_mixed_analysis(full_dict, code, n_classes=k, max_iter=max_iter)
        cluster_df[f"cluster_k{k}"] = labels["cluster"]

        # Evaluación
        pi = model_params["pi"]
        Nk = (pi * len(df)).astype(int)

        resp_probs = labels["cluster"].value_counts(normalize=True).values
        entropy = -np.sum(resp_probs * np.log(resp_probs + 1e-12))

        rows.append({
            "k": k,
            "n_clusters": len(pi),
            "entropy": entropy,
            "min_cluster_size": Nk.min(),
            "max_cluster_size": Nk.max(),
            "count_limited": count_limited,
            "count_non_limited": count_non_limited
        })

    summary_df = pd.DataFrame(rows)

    return cluster_df, summary_df

def evaluate_lca_allk(df, k_list=[2,3,4,5,6]):
    cluster_cols = [f"cluster_k{k}" for k in k_list if f"cluster_k{k}" in df.columns]
    feature_cols = [c for c in df.columns if c not in cluster_cols]

    X = df[feature_cols].copy()
    X = X.apply(pd.to_numeric, errors="coerce")
    X = X.dropna(axis=1, how="all")
    X = X.fillna(X.mean())

    X_scaled = StandardScaler().fit_transform(X.values)

    embedding = PCA(n_components=2, random_state=0).fit_transform(X_scaled)

    df_out = df.copy()
    df_out["pca_x"] = embedding[:, 0]
    df_out["pca_y"] = embedding[:, 1]

    metrics = []

    for k in k_list:
        col = f"cluster_k{k}"
        if col not in df.columns:
            continue

        labels = df[col].values
        unique_labels = np.unique(labels)

        if len(unique_labels) < 2:
            metrics.append({
                "k": k,
                "silhouette": np.nan,
                "calinski_harabasz": np.nan,
                "davies_bouldin": np.nan,
                "n_clusters": 1,
                "collapsed": True
            })
            continue

        sil = silhouette_score(X_scaled, labels)
        ch  = calinski_harabasz_score(X_scaled, labels)
        db  = davies_bouldin_score(X_scaled, labels)

        metrics.append({
            "k": k,
            "silhouette": sil,
            "calinski_harabasz": ch,
            "davies_bouldin": db,
            "n_clusters": len(unique_labels),
            "collapsed": False
        })

    metrics_df = pd.DataFrame(metrics)

    return df_out, metrics_df

def plotly_lca_pca_dropdown(full_dict, code_list, k_list=[2,3,4,5,6]):
    import numpy as np
    import pandas as pd
    from sklearn.preprocessing import StandardScaler
    from sklearn.decomposition import PCA
    import plotly.graph_objects as go

    traces = []
    buttons = []
    trace_map = {}
    trace_idx = 0

    for code in code_list:
        df = full_dict[code]["df_cluster_full"]

        cluster_cols = [f"cluster_k{k}" for k in k_list if f"cluster_k{k}" in df.columns]
        feature_cols = [c for c in df.columns if c not in cluster_cols]

        X = df[feature_cols].apply(pd.to_numeric, errors="coerce")
        X = X.dropna(axis=1, how="all")
        X = X.fillna(X.mean())

        Xs = StandardScaler().fit_transform(X.values)
        emb = PCA(n_components=2, random_state=0).fit_transform(Xs)

        for k in k_list:
            col = f"cluster_k{k}"
            if col not in df.columns:
                continue

            labels = df[col].values
            uniq = np.unique(labels)

            ids = []
            for u in uniq:
                mask = labels == u
                traces.append(go.Scatter(
                    x=emb[mask,0],
                    y=emb[mask,1],
                    mode="markers",
                    marker=dict(size=6),
                    name=f"{code} – k={k} – C{u}",
                    visible=False
                ))
                ids.append(trace_idx)
                trace_idx += 1

            trace_map[(code,k)] = ids

    n = len(traces)

    # dropdown with all combinations
    for (code,k), ids in trace_map.items():
        visible = [False]*n
        for i in ids:
            visible[i] = True

        buttons.append(dict(
            label=f"{code} | k={k}",
            method="update",
            args=[{"visible": visible},
                  {"title": f"{code} – k={k}"}]
        ))

    # default
    first = list(trace_map.keys())[0]
    for i in trace_map[first]:
        traces[i].visible = True

    fig = go.Figure(traces)

    fig.update_layout(
        updatemenus=[dict(buttons=buttons, x=0.01, y=1.15)],
        title=f"{first[0]} – k={first[1]}",
        height=700
    )

    fig.show()
```

# Clusterización
## LCA

Se planteó inicialmente un tipo de clusterización llamada LCA, un modelo probabilístico que asume que la población está formada por $K$ clases latentes y que las variables observadas son codicionalmente independientes dado el cluster. 

### Modelo

Para un individuo $i$ con respuestas $(x_{i1}$, $\dots$, $x_{ip})$:


$P(x_i) = \sum_{k=1}^{K} \pi_k \prod_{j=1}^{p} P(x_{ij} \mid z_i = k)$




$
\sum_{k=1}^{K} \pi_k = 1, \quad \pi_k \ge 0
$

Es decir:
 - La probabilidad de observar sus respuestas $x_i$ es la suma de probabilidades sobre todas las posibles clases latentes $k$.  
- $\pi_k$ = **prior o peso del cluster $k$** (qué tan grande es la clase en la población).  
- $P(x_{ij} \mid z_i=k) = \theta_{jkc}$ = probabilidad de la respuesta de la variable $j$ dado el cluster $k$.  
- Sirve para modelar la **mezcla de perfiles** en la población.

Por otro lado el posterior, que es la probabilidad de que el individuo $i$ pertenezca a la clase $k$ **después de observar sus respuestas**.  
- Sirve para asignar individuos a clusters de manera **probabilística**, no rígida.

Se define así:

$\gamma_{ik} = P(z_i = k \mid x_i) =
\frac{\pi_k \prod_j P(x_{ij} \mid k)}
{\sum_{l=1}^{K} \pi_l \prod_j P(x_{ij} \mid l)}$

### Estimación EM (Expectation-Maximization)

Se usa porque $z_i$ no se observa (es latente). Se hace iterativamente:

**E-step (Expectation)**

$$
\gamma_{ik} = P(z_i = k \mid x_i)
$$

- Calcula la **expectativa de pertenencia** de cada individuo a cada clase según los parámetros actuales.  

**M-step (Maximization)**

$$
\pi_k = \frac{1}{N}\sum_{i=1}^{N} \gamma_{ik}
$$

$$
\theta_{jkc} = \frac{\sum_{i=1}^{N} \gamma_{ik} \mathbf{1}[x_{ij}=c]}{\sum_{i=1}^{N} \gamma_{ik}}
$$

- Actualiza los parámetros del modelo ($\pi_k$, $\theta_{jkc}$) usando las **probabilidades de pertenencia calculadas**.  
- Repite E y M hasta que cambien poco los parámetros → **convergencia**.

### Intuición global

- **$\pi_k$** → “qué tan grande es cada clase en la población” (prior).  
- **$\theta_{jkc}$** → “cómo se comporta cada clase en cada variable”.  
- **$\gamma_{ik}$** → “qué tan probable es que cada individuo pertenezca a cada clase” (posterior).  
- **EM** → algoritmo que ajusta $\pi_k$ y $\theta_{jkc}$ usando $\gamma_{ik}$ iterativamente.  


## LCA para datos mixtos (numéricas + categóricas)

A pesar del planteamiento inicial, es necesario considerar que algunas de estos data subsets contemplan datos numéricos discretos, por ejemplo, en el caso del diagnóstico de bulimia se poseen datos relacionados al peso que dentro de una clusterización deben ser considerados

Para ello, en este caso particular se hacen los siguientes ajustes:


- Variables numéricas: $X_i^\text{num} \in \mathbb{R}^{D_n}$  
- Variables categóricas: $X_i^\text{cat} \in \{0,1,\dots,m_j-1\}^{D_c}$

El modelo asume independencia condicional dentro de cada tipo de variable:

$$
P(x_i) = \sum_{k=1}^{K} \pi_k \, P(X_i^\text{num} \mid z_i=k) \, P(X_i^\text{cat} \mid z_i=k)
$$

- Para las numéricas se usa una **gaussiana multivariante**:

$$
X_i^\text{num} \mid z_i=k \sim \mathcal{N}(\mu_k, \Sigma_k)
$$

- Para las categóricas se usa una **multinomial**:

$$
P(X_{ij}^\text{cat}=c \mid z_i=k) = \theta_{jkc}
$$

---

### Posterior de pertenencia (E-step)

Se calcula combinando ambas distribuciones:

$$
\gamma_{ik} = P(z_i=k \mid x_i) =
\frac{\pi_k \, \mathcal{N}(X_i^\text{num} \mid \mu_k, \Sigma_k) \prod_{j=1}^{D_c} \theta_{jk, X_{ij}^\text{cat}}}
{\sum_{l=1}^{K} \pi_l \, \mathcal{N}(X_i^\text{num} \mid \mu_l, \Sigma_l) \prod_{j=1}^{D_c} \theta_{jl, X_{ij}^\text{cat}}}
$$

- $\gamma_{ik}$ ahora depende de la **densidad de las variables numéricas** y de la **probabilidad categórica**.

---

### Actualización de parámetros (M-step)

- Priors:

$$
\pi_k = \frac{1}{N} \sum_{i=1}^{N} \gamma_{ik}
$$

- Numéricas:

$$
\mu_k = \frac{\sum_{i=1}^{N} \gamma_{ik} X_i^\text{num}}{\sum_{i=1}^{N} \gamma_{ik}}, 
\quad
\Sigma_k = \frac{\sum_{i=1}^{N} \gamma_{ik} (X_i^\text{num}-\mu_k)(X_i^\text{num}-\mu_k)^T}{\sum_{i=1}^{N} \gamma_{ik}}
$$

- Categóricas:

$$
\theta_{jkc} = \frac{\sum_{i=1}^{N} \gamma_{ik} \mathbf{1}[X_{ij}^\text{cat}=c]}{\sum_{i=1}^{N} \gamma_{ik}}
$$

---

### Diferencias clave con LCA estándar

| Aspecto | LCA estándar | LCA mixto |
|---------|-------------|-----------|
| Variables | Solo categóricas | Categóricas + numéricas |
| Distribución | Multinomial | Multinomial + Gaussiana multivariante |
| Parámetros | $\pi_k$, $\theta_{jkc}$ | $\pi_k$, $\theta_{jkc}$, $\mu_k$, $\Sigma_k$ |
| Posterior | $\gamma_{ik} \propto \pi_k \prod_j \theta_{jkc}$ | $\gamma_{ik} \propto \pi_k \, \mathcal{N}(X_i^\text{num}) \prod_j \theta_{jkc}$ |
| EM | Actualiza priors y probabilidades categóricas | Actualiza priors, medias/covarianzas y probabilidades categóricas |

---

### Intuición global

- **$\pi_k$** → “qué tan grande es cada clase en la población”.  
- **$\mu_k$, $\Sigma_k$** → perfil de las variables numéricas en cada clase.  
- **$\theta_{jkc}$** → perfil de las variables categóricas en cada clase.  
- **$\gamma_{ik}$** → probabilidad de pertenencia de cada individuo a cada cluster.  
- **EM** → ajusta todos los parámetros iterativamente usando las responsabilidades $\gamma_{ik}$.

---

## Justificación de las decisiones en LCA mixto

1. **Separación de variables en numéricas y categóricas**  
   - Los datos incluyen variables numéricas y variables categóricas (previo a análisis clínico, es decir, previo a determinación experta).  
   - Tratar todas las variables como categóricas haría perder información de la magnitud y dispersión de las variables numéricas.  
   - Por eso se definen:
     $$
     X_i^\text{num} \in \mathbb{R}^{D_n}, \quad X_i^\text{cat} \in \{0,1,\dots,m_j-1\}^{D_c}
     $$

2. **Uso de una gaussiana multivariante para variables numéricas**  
   - Permite modelar la **media y la covarianza** de las variables numéricas dentro de cada cluster.  
   - Así, clusters similares se agrupan no solo por categorías, sino también por **valores cercanos en las variables continuas**:
     $$
     X_i^\text{num} \mid z_i=k \sim \mathcal{N}(\mu_k, \Sigma_k)
     $$

3. **Uso de distribución multinomial para variables categóricas**  
   - Mantiene la formulación clásica de LCA para las variables categóricas.  
   - Cada cluster tiene un **perfil de probabilidades** para cada categoría de cada variable:
     $$
     P(X_{ij}^\text{cat}=c \mid z_i=k) = \theta_{jkc}
     $$

4. **Posterior de pertenencia combinado**  
   - Al calcular $\gamma_{ik}$ se integran ambas distribuciones, numérica y categórica, para reflejar toda la información disponible de cada individuo:
     $$
     \gamma_{ik} \propto \pi_k \, \mathcal{N}(X_i^\text{num} \mid \mu_k, \Sigma_k) \prod_{j=1}^{D_c} \theta_{jk, X_{ij}^\text{cat}}
     $$
   - Esto garantiza que la asignación a clusters se basa tanto en patrones continuos como discretos.

5. **Actualización de parámetros en M-step**  
   - Priors $\pi_k$ reflejan el tamaño relativo de cada cluster.  
   - $\mu_k$ y $\Sigma_k$ resumen la tendencia central y dispersión de variables numéricas en el cluster.  
   - $\theta_{jkc}$ resume la frecuencia de categorías dentro de cada cluster.  
   - La combinación de estos parámetros permite que cada cluster tenga un **perfil completo** de datos mixtos.

6. **Iteración EM**  
   - La decisión de usar EM se mantiene porque los clusters son **latentes** y no se conocen directamente.  
   - Permite **refinar probabilidades de pertenencia y parámetros simultáneamente**, integrando información numérica y categórica.

---

En resumen, estas decisiones permiten que la clusterización capture:

- **Información cuantitativa** (valores numéricos y su variabilidad).  
- **Información cualitativa** (respuestas categóricas).  
- **Probabilidades de pertenencia suaves** que reflejan la incertidumbre del cluster.  
- Una solución consistente y interpretable para datos clínicos mixtos, que no sería posible usando LCA estándar solo para categóricas.

## Cantidad de Clusters $k$

Para determinar la cantidad de clusteres k es posible considerar un valor general que aplique a todos los diagnósticos, por ejemplo si se ignoraran posibles outliers, 4 clases podrían ser aceptables si se plantea que todos poseen una clase que no posee el diagnóstico y 3 clases que sí con diferenciación para un diagnóstico leve, moderado y grave.


De todas formas, incluso si se definiera un único valor para k, es útil evaluar todos los posibles, siguiendo la lógica anterior, si incluyéramos más granularidad, podemos aumentar hasta dos clases entre leve - moderado y moderado - grave. Así, se evalúa desde una clusterización binaria hasta una de 6 clases.

```python
code_list = [
    'pso', 'pga', 'psa', 'pea', 'pmd', 'pma', 'psz',
    'pad', 'pcd', 'pod', 'pal', 'pmj', 'psu', 'pni'
]


all_summaries = {}

for code in code_list:
    print(full_dict[code]["nombre"], "-", code)
    
    # Ejecuta LCA y obtiene todos los clusters
    cluster_df, summary = lca_evaluate_clusters(full_dict, code, max_k=6, max_iter=100)
    
    # Guardar todos los clusters por k
    full_dict[code]['clustering_df_allk'] = cluster_df
    
    # Guardar summary
    all_summaries[code] = summary
    print(f"\nSummary for {code}:")
    display(summary)
    
    # Crear df original + clusters
    df_orig = full_dict[code]["df"].reset_index(drop=True)
    cluster_df_reset = cluster_df.reset_index(drop=True)
    full_dict[code]["df_cluster_full"] = pd.concat([df_orig, cluster_df_reset], axis=1)

# Exportar CSV
out_dir_csv = Path.cwd() / "resultados" / "clustering" / "csv"
out_dir_csv.mkdir(parents=True, exist_ok=True)
for code in code_list:
    full_dict[code]["df_cluster_full"].to_csv(out_dir_csv / f"{code}_cluster_full.csv", index=False)

# Exportar Excel
out_dir_xlsx = Path.cwd() / "resultados" / "clustering" / "xlsx"
out_dir_xlsx.mkdir(parents=True, exist_ok=True)
for code in code_list:
    full_dict[code]["df_cluster_full"].to_excel(out_dir_xlsx / f"{code}_cluster_full.xlsx", index=False)
```
